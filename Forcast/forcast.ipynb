```python
import pandas as pd
import jdatetime

df1 = pd.read_excel("report (4).xlsx")
df2 = pd.read_excel("report (6).xlsx")

def jalali_to_gregorian(date_str):

    try:
        y, m, d = map(int, str(date_str).split('/'))

        return jdatetime.date(
        y,
        m,
        d
        ).togregorian()

    except:
        return None

# تبدیل تاریخ شمسی فایل دوم

df2['زمان درمان'] = df2['زمان درمان'].apply(
jalali_to_gregorian
)

# حذف ساعت از فایل اول

df1['زمان درمان'] = pd.to_datetime(
df1['زمان درمان']
).dt.date

# تبدیل فایل دوم به date

df2['زمان درمان'] = pd.to_datetime(
df2['زمان درمان']
).dt.date

print(df1['زمان درمان'].head())
print(df2['زمان درمان'].head())
```

```python
monthly_income = merged_df.resample(
'ME',
on='زمان درمان'
)['سهم پزشک'].sum().reset_index()
```

```python
monthly_income.columns = ['ds', 'y']

print(monthly_income.head())
```

```python
from prophet import Prophet
```

```python
model = Prophet()

model.fit(monthly_income)
```

```python
future = model.make_future_dataframe(
periods=6,
freq='ME'
)
```

```python
forecast = model.predict(future)

print(
forecast[
['ds', 'yhat']
].tail(6)
)
```

```python
from prophet.diagnostics import cross_validation, performance_metrics

# این دستور مدل را در بازه‌های مختلف تست می‌کند
# معمولاً برای داده‌های ماهانه، بازه horizon را ۱۸۰ روز (۶ ماه) می‌گذارند
df_cv = cross_validation(model, initial='365 days', period='30 days', horizon='180 days')

# محاسبه متریک‌های دقت
df_p = performance_metrics(df_cv)
print(df_p[['horizon', 'mae', 'rmse', 'mape']].head())
```

```python
# ۱. حذف مقادیر منفی (که احتمالاً اصلاحیه یا برگشت وجه هستند)
monthly_income_clean = monthly_income[monthly_income['y'] > 0]

# ۲. حذف داده‌های خیلی پرت (بیشتر از ۳ برابر انحراف معیار - اختیاری اما توصیه شده)
mean_y = monthly_income_clean['y'].mean()
std_y = monthly_income_clean['y'].std()
# نگه داشتن داده‌هایی که در بازه منطقی هستند
monthly_income_clean = monthly_income_clean[(monthly_income_clean['y'] > (mean_y - 2*std_y))]

print(f"تعداد داده‌های باقی‌مانده: {len(monthly_income_clean)}")
```

```python
# ۱. تعیین سقف منطقی برای درآمد ماهانه (مثلاً ۵ میلیارد تومان)
# هر عددی بالاتر از این به عنوان نوسان مخرب حذف می‌شود
upper_limit = 5_000_000_000
lower_limit = 50_000_000   # حذف درآمدهای نزدیک به صفر یا منفی

# ۲. فیلتر کردن داده‌های پرت
df_cleaned = monthly_income[
    (monthly_income['y'] <= upper_limit) &
    (monthly_income['y'] >= lower_limit)
].copy()

# ۳. بازسازی مدل با پارامترهای متعادل‌تر
m = Prophet(
    seasonality_mode='additive', # تغییر به additive برای جلوگیری از درصدهای نجومی
    changepoint_prior_scale=0.01, # کاهش حساسیت برای نرم‌تر شدن خط روند
    yearly_seasonality=True,
    interval_width=0.95 # بازه اطمینان ۹۵ درصد
)

m.fit(df_cleaned)

# ۴. پیش‌بینی ۱۲ ماه آینده
future = m.make_future_dataframe(periods=12, freq='M')
forecast = m.predict(future)

# ۵. رسم نمودار با اعمال محدودیت در نمایش برای دیدن نقاط سیاه
import matplotlib.pyplot as plt

fig = m.plot(forecast)
ax = fig.gca()
ax.set_ylim(0, 6_000_000_000) # سقف نمایش نمودار روی ۶ میلیارد تومان
ax.set_title("پیش‌بینی بهینه شده (بدون داده‌های پرت)")
plt.show()

# ۶. رسم اجزای مدل (روند و فصل)
m.plot_components(forecast)
plt.show()
```

```python
import matplotlib.pyplot as plt
from prophet import Prophet

# ۱. فیلتر کردن داده‌های پرت (حذف مبالغ غیرمنطقی بالای ۵ میلیارد و زیر ۵۰ میلیون)
upper_limit = 5_000_000_000
lower_limit = 50_000_000

df_cleaned = monthly_income[
    (monthly_income['y'] <= upper_limit) &
    (monthly_income['y'] >= lower_limit)
].copy()

# ۲. تعریف مدل با تنظیمات اصلاح شده
# استفاده از additive برای جلوگیری از درصدهای نجومی در نمودار فصلی
m = Prophet(
    seasonality_mode='additive',
    changepoint_prior_scale=0.01,
    yearly_seasonality=True
)

m.fit(df_cleaned)

# ۳. پیش‌بینی ۱۲ ماه آینده با استفاده از فرکانس اصلاح شده 'ME'
# توجه: 'M' منسوخ شده و بجای آن از 'ME' استفاده می‌کنیم
future = m.make_future_dataframe(periods=12, freq='ME')
forecast = m.predict(future)

# ۴. ترسیم نمودار اصلی و تنظیم محورها برای دیده شدن نقاط سیاه
fig = m.plot(forecast)
ax = fig.gca()
ax.set_ylim(0, 6_000_000_000) # محدود کردن سقف برای وضوح نقاط سیاه
ax.set_title("Forecast with Month-End (ME) Frequency")

# ۵. رسم نمودار اجزاء (روند و فصلی بودن)
m.plot_components(forecast)
plt.show()
```

```python
# ۱. محاسبه سقف بر اساس میانگین داده‌ها (روش هوشمند)
# این کار داده‌هایی که بیش از ۳ برابر میانگین هستند را حذف می‌کند
median_income = monthly_income['y'].median()
dynamic_upper_limit = median_income * 3

df_cleaned = monthly_income[
    (monthly_income['y'] <= dynamic_upper_limit) &
    (monthly_income['y'] >= 50_000_000)
].copy()

# ۲. بازسازی مدل با کنترل شدید روی نوسانات فصلی
m = Prophet(
    seasonality_mode='additive',
    yearly_seasonality=10, # فیکس کردن تعداد دوره‌های فصلی برای نرم‌تر شدن نمودار
    changepoint_prior_scale=0.01
)

m.fit(df_cleaned)

# ۳. پیش‌بینی ۱۲ ماهه با ME
future = m.make_future_dataframe(periods=12, freq='ME')
forecast = m.predict(future)

# ۴. رسم نمودار با مقیاس خوانا (میلیارد تومان)
import matplotlib.ticker as ticker

fig = m.plot(forecast)
ax = fig.gca()
# تبدیل اعداد محور Y به میلیارد تومان برای خوانایی
ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{x/1e9:.1f}B'))
ax.set_title("Predicted Revenue (Billions)")

m.plot_components(forecast)
plt.show()
```

```python
# ترسیم اجزای مدل
fig2 = m.plot_components(forecast)

# اصلاح مقیاس محور Y در نمودار فصلی (Yearly) برای حذف 1e11
# این کد تمام زیرنمودارها را بررسی و اعداد را به میلیارد تبدیل می‌کند
for ax in fig2.get_axes():
    ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{x/1e9:.1f}B'))

plt.show()
```

```python
# استخراج پیش‌بینی ۱۲ ماه آینده (فقط ماه‌های آتی)
final_report = forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail(12)

# تبدیل به میلیارد برای خوانایی در اکسل
final_report[['yhat', 'yhat_lower', 'yhat_upper']] = final_report[['yhat', 'yhat_lower', 'yhat_upper']] / 1e9

# ذخیره نهایی
final_report.to_excel("Final_Revenue_Forecast_Billion.xlsx", index=False)
print("گزارش نهایی با واحد میلیارد تومان ذخیره شد.")
```
